## 2.1 Dasar Teori

2.1 Pengertian Perulangan for

Perulangan for pada Python mengulang satu blok instruksi untuk setiap elemen dalam suatu urutan. Setiap kali blok dijalankan, variabel perulangan (misalnya i) menyimpan nilai elemen yang sedang diproses. Blok yang diulang ditandai dengan indentasi (menjorok ke dalam) di bawah baris for. Perulangan berhenti otomatis setelah seluruh elemen urutan diproses.

Pada latihan pertama, jumlah pengulangan diambil dari masukan pengguna yang dikonversi ke bilangan bulat, lalu dipakai sebagai batas perulangan. Hasilnya adalah keterangan "Perulangan ke-i" yang tercetak sebanyak nilai masukan. Ini menunjukkan bahwa banyaknya pengulangan dapat ditentukan secara dinamis saat program berjalan.

2.2 Fungsi range()

Fungsi range() menghasilkan urutan bilangan bulat yang dilewati oleh perulangan for. Range(stop) :	mulai dari 0 sampai stop - 1, contoh: 0 1 2 3 4. range(start, stop)	mulai dari start sampai stop - 1	contoh: 1 2 3 4 5. Range(start, stop, step)	naik atau turun dengan langkah step, contoh: 4 3 2 1. Hal yang perlu diperhatikan adalah nilai stop tidak ikut diproses.Dengan masukan 10, bentuk satu argumen mencetak 0 sampai 9, sedangkan bentuk dua argumen dengan batas masukan + 1 mencetak 1 sampai 10. Jadi, untuk memproses bilangan 1 sampai N, batas atasnya harus N + 1. 

2.3 Parameter end pada print()

Secara bawaan, print() mengakhiri keluarannya dengan karakter baris baru sehingga setiap pemanggilan tampil di baris tersendiri. Parameter end mengubah perilaku ini. Dengan end=" ", keluaran berikutnya tampil di baris yang sama dan dipisahkan spasi. Dengan end="", keluaran berikutnya menempel langsung tanpa pemisah.

Pada latihan deret angka, parameter ini dipakai agar bilangan tampil berjajar dalam satu baris. Pemanggilan print() tanpa argumen setelah perulangan dipakai untuk pindah ke baris berikutnya. Pasangan end dan print() kosong ini menjadi dasar pembuatan pola karena satu baris pola dirakit dari banyak potongan karakter.

2.4 Akumulator dan Percabangan di dalam Perulangan

Latihan penjumlahan deret memperlihatkan penggunaan akumulator, yaitu variabel yang diberi nilai awal 0 sebelum perulangan, lalu ditambah nilai baru pada setiap iterasi. Setelah perulangan selesai, variabel tersebut berisi total seluruh bilangan yang diproses.

Latihan ini juga memakai percabangan if-else di dalam perulangan. Selama bilangan yang diproses belum mencapai batas, program mencetak tanda tambah sebagai penghubung antarbilangan. Pada bilangan terakhir, program mencetak tanda sama dengan beserta hasil penjumlahan. Dengan demikian, perulangan dapat menghasilkan perilaku yang berbeda pada iterasi terakhir dibanding iterasi lainnya, dan keluarannya terbaca sebagai persamaan matematika yang utuh.

2.5 Perulangan Bersarang (Nested Loop)

Perulangan bersarang adalah perulangan for yang berada di dalam perulangan for lain. Setiap satu kali iterasi perulangan luar, perulangan dalam dijalankan sampai selesai. Jika perulangan luar berjalan sebanyak m kali dan perulangan dalam sebanyak n kali, blok terdalam dijalankan sebanyak m × n kali.

Dalam pembuatan pola, pembagian tugasnya adalah sebagai berikut: perulangan luar mengatur baris, sedangkan perulangan dalam mengatur kolom atau isi pada baris tersebut. Pemanggilan print() kosong ditempatkan setelah perulangan dalam agar kursor pindah ke baris berikutnya.

Pengaruh batas perulangan dalam terhadap bentuk pola tampak pada beberapa latihan:

Batas tetap (persegi bintang): perulangan dalam selalu berjalan sebanyak jumlah yang sama untuk setiap baris, sehingga jumlah karakter tiap baris sama dan bentuknya persegi.
Batas bergantung variabel luar (pola titik dan angka bertingkat): jumlah pengulangan dalam berubah mengikuti nomor baris, sehingga jumlah titik berkurang di tiap baris sementara angka di ujung baris bertambah.
Nilai bergantung indeks baris dan kolom (tabel penjumlahan): setiap sel bernilai jumlah variabel baris dan kolom, sehingga tiap baris tampil bergeser satu angka dari baris sebelumnya.

2.6 Percabangan Berlapis di dalam Nested Loop

Pada latihan pola bingkai, karakter yang dicetak ditentukan oleh posisi (baris, kolom) saat itu melalui rangkaian if-elif-else. Baris pertama dan terakhir dicetak sebagai bingkai horizontal, sedangkan kolom pertama dan terakhir dicetak sebagai garis tegak. Di dalam bingkai, posisi tertentu dicetak sebagai tanda < dan >, posisi di antaranya dicetak sebagai titik, dan sisanya dicetak sebagai spasi. Posisi tanda tersebut digeser di setiap baris melalui variabel bantu yang nilainya diperbarui pada akhir baris. Pendekatan ini memperlihatkan bahwa perulangan bersarang dapat dipadukan dengan percabangan untuk menentukan isi setiap sel secara individual.

2.7 Penerapan Input Dinamis dan Penamaan Variabel

Seluruh program pada praktikum menerima ukuran atau batas dari pengguna melalui input(), lalu diubah ke bilangan bulat dengan int(). Karena input() selalu menghasilkan string, konversi ini wajib agar nilai dapat dipakai sebagai batas range() dan dalam perhitungan. Selain itu, seluruh variabel diberi akhiran 4 digit NIM sebagai identitas pengerjaan mandiri. Penamaan yang bermakna, seperti baris, spasi, dan angka, juga membuat alur program lebih mudah dipahami.

2.2.4 Program "nested_for_1_2611531001"

```python
batas_1001 =int(input("Masukkan nilai batas: "))
for line_1001 in range(1, batas_1001+1):
    for j in range(1, (-1 * line_1001 + batas_1001) + 1):
        print("." \
        "1", end= "")
    print(line_1001)
```

   Loop dalam untuk menghitung berapa banyak titik yang dicetak. 
   
3. Rumus (-1 * line_1001 + batas_1001):

2.2.7 Program "nested_for_4_2611531001"

```python
tinggi_1001 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1001 % 2 !=0:
    print("Tinggi harus bilangan genap!")
else:
    a_1001 = tinggi_1001
    c_1001 = a_1001
    lebar_1001 = (2 * tinggi_1001) - 2

    for i_1001 in range(1, tinggi_1001 + 1):
        b_1001 = c_1001 + 1

        for j_1001 in range(1, lebar_1001 + 1):

            #Baris atas dan bawah
            if i_1001 == 1 or i_1001 == tinggi_1001:
                if j_1001 == 1 or j_1001 == lebar_1001:
                    print("#", end="")
                else:
                    print("=", end="")

            #Baris isi
            else:
                if j_1001 == 1 or  j_1001 == lebar_1001:
                    print("|", end="")
                else:
                    if j_1001 == c_1001:
                        print("<", end="")
                    elif j_1001 == b_1001:
                        print(">", end="")
                    elif j_1001 == (lebar_1001 - c_1001):
                        print("<", end="")
                    elif j_1001 == (lebar_1001 - c_1001 + 1):
                        print(">", end="")
                    elif j_1001 > b_1001 and j_1001< (lebar_1001 - c_1001):
                        print(".", end="")
                    else:
                        print(" ", end="")
        print()

        # Logika asli Java
        a_1001 -= 2

        if a_1001 <= 0:
            c_1001 = (-a_1001) +2
        else:
            c_1001 = a_1001
```

Penjelasan:

Kode program pada ketiga gambar tersebut merupakan satu kesatuan program Python lengkap untuk mencetak pola simetris berbentuk bingkai dengan karakter dekoratif di dalamnya.

1. tinggi_1001 = int(...): 

   Meminta input angka dari pengguna sebagai batas tinggi pola.
   
   if tinggi_1001 % 2 != 0:
   
   Memeriksa apakah input adalah bilangan ganjil. Jika ganjil, program mencetak pesan "Tinggi harus bilangan genap!" dan berhenti.
   
2. a_1001 = tinggi_1001 & c_1001 = a_1001: Mengatur nilai variabel kontrol posisi.   

   lebar_1001 = (2 * tinggi_1001) - 2: Menghitung lebar total pola berdasarkan tinggi yang diinputkan.

  3. Perulangan Utama for i_1001 in range(1, tinggi_1001 + 1): Loop luar untuk mengontrol baris dari atas ke bawah.

      b_1001 = c_1001 + 1: Menentukan posisi kolom pendamping.   

      for j_1001 in range(1, lebar_1001 + 1): Loop dalam untuk mengontrol kolom dari kiri ke kanan.  

 4. Bingkai Atas dan bawah(baris 15-20)

    A. baris pertama (i_1001 == 1) dan baris terakhir (i_1001 == tinggi_1001), kolom paling ujung kiri dan kanan dicetak #, sedangkan kolom di antaranya dicetak =.

    B. Pembatas Samping (Baris 23–25)Pada baris isi (tengah), kolom paling kiri dan kanan dicetak |.     
    
    C. Isian Dalam Bingkai (Baris 27–38)
    
       < dan >: Dicetak pada koordinat kolom c_1001, b_1001, serta posisi simetris terbaliknya (lebar_1001 - c_1001).
    
       (Titik): Dicetak pada ruang tengah di antara dua kelompok simbol <>.   Spasi Kosong (" "): Dicetak pada area sisa di luar pola.
    
5. print(): Pindah ke baris baru setelah satu baris horizontal selesai dicetak.  

   a_1001 -= 2: Mengurangi nilai variabel kontrol. 
 
   if a_1001 <= 0:: Saat perulangan melewati pertengahan tinggi, posisi c_1001 dibalik nilainya ((-a_1001) + 2). Hal ini membuat bentuk pola bagian bawah menjadi cermin simetris dari bagian atasnya. 

2.2.8 Program "latihan5_2611531001"

```python
tinggi_1001 = int(input("Masukkan tinggi segitiga yang diinginkan: "))

for i_1001 in range(1, tinggi_1001 + 1):
   print(" " * (tinggi_1001 - i_1001), end="")

   print("* " * i_1001)
```

4. print("* " * i_1001):

   Mencetak karakter bintang ber spasi *  sebanyak nilai baris aktif (i_1001).

```python
print("=== PROGRAM JAM PASIR KRISTAL PALINDROMI ===")
n_1001 = int(input("Masukkan ukuran skala jam pasir (N): "))

# Bingkai atas 
print("#", end="")
for i_1001 in range(4 * n_1001 + 5):
    print("=", end="")
print("#")

#-------Fase 1 Angka menurun (baris N -> 1)-------
for baris_1001 in range(n_1001, 0, -1):
    print("| ", end="")
    for spasi_1001 in range(2 * (n_1001 - baris_1001)):
        print(" ", end="")
    for angka_1001 in range(baris_1001, 0, -1):
        print(angka_1001, end=" ")
    print("<*>", end="")
    for angka_1001 in range(1, baris_1001 + 1):
        print(" ", end="")
        print(angka_1001, end="")
    for spasi_1001 in range(2 * (n_1001 - baris_1001)):
        print(" ", end="")
    print(" |", end="")
    print()

#--------Fase 2 Poros titik pusat-------
print("|", end="")
for spasi_1001 in range(2 * n_1001 + 1):
    print(" ", end="")
print("<*>", end="")
for spasi_1001 in range(2 * n_1001 + 1):
    print(" ", end="")
print("|", end="")
print()

#--------Fase 3 Angka menaik (baris 1 -> N)------
for baris_1001 in range(1, n_1001 + 1):
    print("| ", end="")
    for spasi_1001 in range(2 * (n_1001 - baris_1001)):
        print(" ", end="")
    for angka_1001 in range(baris_1001, 0, -1):
        print(angka_1001, end=" ")
    print("<*>", end="")
    for angka_1001 in range(1, baris_1001 + 1):
        print(" ", end="")
        print(angka_1001, end="")
    for spasi_1001 in range(2 * (n_1001 - baris_1001)):
        print(" ", end="")
    print(" |", end="")
    print()

# ---------- Bingkai bawah ----------
print("#", end="")
for i_1001 in range(4 * n_1001 + 5):
    print("=", end="")
print("#")
```

Penjelasan:

Program ini bertujuan untuk mencetak pola angka berbentuk jam pasir yang palindromik (simetris kiri-kanan dan atas-bawah) di dalam sebuah bingkai bertema kristal. 

1. Input Skala dan Bingkai Atas (Baris 1–8)

   a. Baris 1–2: Menampilkan judul program dan meminta input nilai n_1001 (skala ukuran jam pasir) dari pengguna.
   b. Baris 5–8: Mencetak tutup bingkai bagian atas. Sudut kiri dan kanan diisi karakter #, sedangkan bagian tengahnya diisi garis horizontal = sebanyak 4 * n_1001 + 5 kali.
   
2. Paruh Atas Jam Pasir: Angka Menurun (Baris 10–24)

   Perulangan utama baris_1001 berjalan secara mundur (decrement) dari n_1001 turun ke 1:
   
   a. Baris 12: Mencetak pembatas kiri bingkai | .
   
   b. Baris 13–14: Mencetak spasi awal yang semakin banyak seiring mengecilnya ukuran deret angka, membuat pola menyempit ke tengah.

   c. Baris 15–16: Mencetak deret angka bagian kiri yang menurun dari baris_1001 ke 1.
   
   d. Baris 17: Mencetak manik Kristal Pusat <*> tepat di tengah.
   
   e. Baris 18–20: Mencetak deret angka bagian kanan yang menaik dari 1 ke baris_1001.
   
   f. Baris 21–22: Mencetak spasi penyeimbang di sisi kanan.
   
   g. Baris 23–24: Mencetak pembatas kanan bingkai |, lalu pindah ke baris baru.
   
3. Poros Titik Pusat (Baris 26–34)

   Baris ini membentuk garis horizontal tepat di tengah/leher jam pasir:  

   a. Baris 27: Mencetak pembatas kiri |.
   
   b. Baris 28–29: Mencetak spasi kosong sejauh 2 * n_1001 + 1.
   
   c. Baris 30: Mencetak titik tengah <*> tanpa ada deret angka di sampingnya.
   
   d. Baris 31–32: Mencetak spasi kosong penyeimbang di sebelah kanan.
   
   e. Baris 33–34: Mencetak pembatas kanan |, lalu pindah ke baris baru.
   
4. Paruh Bawah Jam Pasir: Angka Menaik (Baris 36–50)

   Perulangan utama baris_1001 berjalan maju dari 1 hingga n_1001:   
   
   a. Struktur cetakannya simetris dengan paruh atas, tetapi jumlah angkanya mulai melebar dari 1 angka hingga kembali ke n_1001 angka.
   
   b. Terdiri dari pembatas kiri | , spasi luar, deret angka menurun (baris_1001 ke 1), kristal pusat <*>, deret angka menaik (1 ke baris_1001), spasi luar, dan diakhiri pembatas kanan |.
   
5. Bingkai Bawah (Baris 52–56)

   a. Baris 53–56: Menutup bingkai bagian bawah dengan format yang persis sama seperti bingkai atas: diawali #, diisi = sebanyak 4 * n_1001 + 5, dan diakhiri #.

# KESIMPULAN

Perulangan for bersama range() memungkinkan pengulangan proses secara terkontrol dan sesuai untuk kasus dengan jumlah pengulangan yang diketahui. Nilai batas akhir range() tidak ikut diproses, sehingga batas atas perlu disesuaikan agar rentang yang diinginkan tercapai.Bentuk range dengan tiga argumen dapat menghasilkan deret naik maupun turun, sehingga satu jenis perulangan dapat dipakai untuk menghitung maju dan mundur. Parameter end dan print() kosong membuat satu baris keluaran dapat dirakit dari banyak potongan. Keduanya menjadi dasar pembuatan pola di konsol.

Akumulator dan percabangan di dalam perulangan memungkinkan penghitungan total serta perlakuan khusus pada iterasi tertentu. Perulangan bersarang membagi tugas dengan jelas: perulangan luar mengatur baris dan perulangan dalam mengatur kolom. Bentuk pola ditentukan oleh hubungan antara batas perulangan dalam dan variabel perulangan luar. Pola Jam Pasir Kristal Palindromik Berbingkai pada tugas 5 dapat diselesaikan tanpa perkalian string. Kerapiannya dijamin oleh perhitungan lebar isi 4N + 5, dan dua bagian jam pasir cukup dibedakan oleh arah perulangan baris.
