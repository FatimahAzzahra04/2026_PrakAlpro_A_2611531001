## 1.1 Latar Belakang
Dalam dunia pemrograman, struktur kontrol percabangan merupakan salah satu konsep fundamental yang wajib dikuasai oleh setiap programmer. Struktur percabangan memungkinkan sebuah program untuk mengambil keputusan dan mengeksekusi blok kode tertentu berdasarkan suatu kondisi yang telah ditentukan. Tanpa adanya struktur percabangan, sebuah program hanya akan berjalan secara linear dari atas ke bawah tanpa mampu merespons berbagai kemungkinan input maupun kondisi yang berbeda-beda dari pengguna.

Bahasa pemrograman Python menyediakan beberapa bentuk struktur percabangan, di antaranya pernyataan if tunggal, if-else, if-elif-else, penggunaan beberapa pernyataan if secara berurutan (multiple if), hingga pernyataan match-case yang diperkenalkan pada Python versi 3.11.9 sebagai alternatif dari switch-case pada bahasa pemrograman lain. Setiap bentuk percabangan tersebut memiliki karakteristik, kegunaan, dan konteks penerapan yang berbeda-beda tergantung pada permasalahan yang ingin diselesaikan.

## 2.1 Dasar Teori
Percabangan (branching/selection) adalah struktur kontrol yang digunakan untuk memilih blok kode yang akan dieksekusi berdasarkan hasil evaluasi suatu kondisi logika (boolean). Berikut adalah bentuk-bentuk struktur percabangan yang digunakan dalam praktikum ini.

 a. Pernyataan if tunggal — digunakan ketika hanya ada satu blok kode yang perlu dijalankan apabila suatu kondisi bernilai benar (True). Jika kondisi bernilai salah (False), blok kode di dalam if akan dilewati begitu saja.

 b. Pernyataan if-else — digunakan ketika terdapat dua kemungkinan hasil, yaitu satu blok kode dijalankan apabila kondisi bernilai True, dan blok kode lain (di dalam else) dijalankan apabila kondisi bernilai False.
 
 c. Pernyataan if-elif-else — digunakan ketika terdapat lebih dari dua kemungkinan kondisi. Python akan mengevaluasi kondisi secara berurutan dari atas ke bawah, dan hanya akan mengeksekusi satu blok kode pertama yang kondisinya bernilai True, kemudian keluar dari struktur percabangan tersebut.
 
 d. Multiple if (beberapa if berdiri sendiri) — berbeda dengan if-elif-else, setiap pernyataan if pada bentuk ini dievaluasi secara independen dan tidak saling berkaitan. Akibatnya, lebih dari satu blok kode dapat dieksekusi sekaligus apabila lebih dari satu kondisi bernilai True.
 
 e. Pernyataan match-case — merupakan struktur pattern matching yang diperkenalkan pada Python 3.10. Nilai suatu variabel dicocokkan dengan pola-pola (case) yang telah didefinisikan secara berurutan, dan blok kode pada case yang cocok pertama kali akan dieksekusi. Case _ berfungsi sebagai kondisi default apabila tidak ada pola yang cocok.
 
 f. Operator logika and — digunakan untuk menggabungkan dua kondisi atau lebih, di mana keseluruhan ekspresi hanya akan bernilai True apabila semua kondisi yang digabungkan bernilai True.

2.2.3 Program "multi_if1_2611531001"

```python
umur_1001 = int(input("Input umur anda: "))
sim_1001 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0]

if umur_1001  >= 17 and sim_1001 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")

if umur_1001 >= 17 and sim_1001 != 'y':
    print("Anda sudah dewasa tapi tidak boleh bawa motor")

if umur_1001 < 17 and sim_1001 != 'y' :
    print("Anda belum cukup umur bawa motor")

if umur_1001 < 17 and sim_1001 == 'y':
    print("Anda belum cukup umur punya SIM")
```

Penjelasan:

1. Input Data dan Parsing (Baris 1-2)

   a. Baris 1 (umur_1001 = int(...)): Mengambil masukan umur dari pengguna dan mengonversinya menjadi bilangan bulat (integer).
   
   b. Baris 2 (sim_1001 = input(...)[0]): Mengambil masukan status SIM, lalu menggunakan indeks [0] untuk mengambil karakter pertama saja (misalnya mengambil 'y' dari jawaban "ya").
   
2. Pengecekan Kondisi dengan Multi-IF (Baris 4–14)
Berbeda dengan struktur if-elif-else, pendekatan di kode ini menggunakan 4 blok if independen yang mutually exclusive (saling pilah):

   a. Baris 4–5: Memeriksa apakah umur >=  17 DAN punya SIM (== 'y'). Jika True, mencetak: "Anda sudah dewasa dan boleh bawa motor".
   
   b. Baris 7–8: Memeriksa apakah umur  >= 17 DAN tidak punya SIM (!= 'y'). Jika True, mencetak: "Anda sudah dewasa tapi tidak boleh bawa motor".
   
   c. Baris 10–11: Memeriksa apakah umur < 17 DAN tidak punya SIM (!= 'y'). Jika True, mencetak: "Anda belum cukup umur bawa motor".
   
   d. Baris 13–14: Memeriksa apakah umur < 17 DAN mengklaim punya SIM (== 'y'). Jika True, mencetak: "Anda belum cukup umur punya SIM".
   
3. Catatan: Meskipun program mengevaluasi keempat blok if satu per satu, hanya satu blok if yang akan bernilai True dan mencetak pesan ke layar karena seluruh kombinasinya sudah mencakup seluruh kemungkinan kondisi.   


2.2.4 Program "multi_if2_2611531001"

```python
# Input dari user
total_belanja_1001 = float(input("Masukkan total belanja (Rp): "))

# Input dari status member ( mengecek apakah user mengetik 'y' atau 'ya')
input_member_1001 = input("Apakah anda member (y/t): ").strip().lower()
is_member_1001 = input_member_1001 in ["y", "ya"]

# Input status kode promo (mengecek apakah user mengetik 'y' atau 'ya')
input_promo_1001 = input("Apakah kode promo valid? (y/t):").strip().lower()
kode_promo_valid_1001 = input_promo_1001 in["y", "ya"]

total_diskon_persen_1001 = 0

if total_belanja_1001 > 10000000:
    total_diskon_persen_1001 +=10 #Diskon belanja besar

if is_member_1001:
    total_diskon_persen_1001 += 5 #Diskon member

if kode_promo_valid_1001:
    total_diskon_persen_1001 +=15 #Diskon voucher

#Menghitung nominal diskon dan total bayar
nominal_diskon_1001 = total_belanja_1001 * (total_diskon_persen_1001 / 100)
total_bayar_1001 = total_belanja_1001 - nominal_diskon_1001

#Output hasil
print("/n--- Rincian Pembayaran ---")
print(f"Total Diskon : {total_diskon_persen_1001}%(Rp{nominal_diskon_1001:,.0f})")
print(f"Total Bayar : Rp {total_bayar_1001: ,.0f}")

print(f"Total diskon yang anda dapatkan: {total_diskon_persen_1001}%")
# Output total diskon yang anda dapatkan : 30% jika belanja > 1juta, member, dan kode promo valid
```

Penjelasan:

1. Input Data dan Validasi String (Baris 1–10)

   a. Baris 2 (total_belanja_1001): Menerima input nominal total belanja dari pengguna dan mengonversinya menjadi angka desimal (float).

   b. Baris 5–6 (is_member_1001): Menerima input status keanggotaan..strip().lower() digunakan untuk menghapus spasi liar dan mengubah huruf menjadi kecil.Operator in ["y", "ya"] mengecek apakah jawaban pengguna adalah "y" atau "ya", lalu menyimpannya sebagai nilai boolean (True atau False).

   c. Baris 9–10 (kode_promo_valid_1001): Melakukan proses pembersihan teks (string handling) dan pengecekan yang sama untuk status validasi kode promo.

2. Akumulasi Diskon Bertingkat dengan Multi-IF (Baris 12–21)

   a. Baris 12 (total_diskon_persen_1001 = 0): Menginisialisasi persentase diskon awal sebesar 0%.

   b. Baris 14–15 (if total_belanja_1001 > 1000000:): Jika total belanja melebihi Rp 1.000.000, persentase diskon bertambah 10%.

   c. Baris 17–18 (if is_member_1001:): Blok if independen (bukan elif) untuk mengecek status member. Jika bernilai True, diskon bertambah lagi 5%.

   d. Baris 20–21 (if kode_promo_valid_1001:): Blok if independen untuk mengecek voucher promo. Jika bernilai True, diskon bertambah lagi 15%.

   e. Catatan Logika: Karena menggunakan if terpisah (bukan elif), seluruh kondisi akan diperiksa satu per satu sehingga semua diskon yang memenuhi syarat dapat terakumulasi (ditumpuk) hingga maksimal 30%.

3. Perhitungan Nominal Pembayaran (Baris 23–25)

   a. Baris 24 (nominal_diskon_1001): Menghitung jumlah potongan harga dalam rupiah dengan rumus:   total_belanja x (total_diskon_persen / 100).

   b. Baris 25 (total_bayar_1001): Menghitung total tagihan akhir dengan mengurangi total belanja awal dengan nominal diskon.

4. Menampilkan Output Rincian Pembayaran (Baris 28–32)

   a. Baris 28–30: Mencetak header rincian pembayaran, total persentase beserta nominal rupiah diskon, dan total biaya yang harus dibayar. Format :, .0f digunakan untuk menampilkan angka desimal tanpa koma dan memberi pemisah ribuan (thousand separator).

   b. Baris 32: Mencetak kembali ringkasan total persentase diskon yang berhasil didapatkan oleh pengguna.







2.2.5 Program "if_elif_else_2611531001"

```python
umur_1001 = int(input("Input umur anda: "))
sim_1001 = input("Apakah Anda Sudah Punya SIM C: ")[0]

if umur_1001 >= 17 and sim_1001 == 'y':
    print("Anda sudah dewasa dan boleh bawa motor")
elif umur_1001 >= 17 and sim_1001 != 'y':
     print("Anda sudah dewasa tetapi tidak boleh bawa motor")
elif umur_1001 < 17 and sim_1001 =='y':
     print("Anda belum cukup umur punya SIM")
else:
     print("Anda belum cukup umur dan tidak boleh bawa motor")
print("Program Selesai")
```

Penjelasan:

1. Baris 1: umur_1001 = int(input("Input umur anda: "))
Mengambil masukan umur dari pengguna melalui input(). Fungsi int(...) mengubah teks masukan menjadi tipe data bilangan bulat (integer) yang kemudian disimpan dalam variabel umur_1001.

2. Baris 2: sim_1001 = input("Apakah Anda Sudah Punya SIM C: ")[0]  
Menerima input status kepemilikan SIM dari pengguna.   Penggunaan [0] (string indexing) berfungsi untuk mengambil hanya karakter pertama dari jawaban pengguna. Misalnya, jika pengguna mengetik "ya", yang diambil dan disimpan ke variabel sim_1001 hanyalah huruf 'y'.

3. Baris 4-5: if umur_1001 >= 17 and sim_1001 == 'y':
Kondisi 1: Memeriksa apakah umur pengguna >= 17 tahun DAN jawaban SIM bernilai 'y'. Jika kedua syarat terpenuhi, program mencetak: "Anda sudah dewasa dan boleh bawa motor".

4. Baris 6-7: elif umur_1001 >= 17 and sim_1001 != 'y':
Kondisi 2: Memeriksa apakah umur pengguna >= 17 tahun DAN jawaban SIM bukan 'y'. Jika syarat ini terpenuhi, program mencetak: "Anda sudah dewasa tetapi tidak boleh bawa motor".

5. Baris 8-9: elif umur_1001 < 17 and sim_1001 == 'y':
Kondisi 3: Memeriksa apakah umur pengguna < 17 tahun DAN jawaban SIM bernilai 'y'. Kondisi ini menangani kasus tidak logis di mana seseorang belum 17 tahun tetapi mengaku punya SIM. Jika syarat terpenuhi, program mencetak: "Anda belum cukup umur punya SIM".

6. Baris 10-11: else:  Kondisi Cadangan: Dijalankan jika seluruh kondisi di atas bernilai False (artinya umur  17 tahun DAN jawaban SIM bukan 'y'). Program akan mencetak: "Anda belum cukup umur dan tidak boleh bawa motor".

7. Baris 12: print("Program Selesai") Berada di luar blok percabangan (tanpa indentasi). Teks "Program Selesai" akan selalu dicetak pada bagian akhir setelah evaluasi kondisi selesai dilakukan.




2.2.6 Program "match_case1_2611531001"

```python
bulan_1001 = int(input("masukkan angka bulan (1-12): "))

match bulan_1001:
    case 1:
        print("Januari")
    case 2:
        print("Februari")
    case 3:
        print("Maret")
    case 4:
        print("April")
    case 5:
        print("Mei")
    case 6:
        print("Juni")
    case 7:
        print("Juli")
    case 8:
        print("Agustus")
    case 9:
        print("September")
    case 10:
        print("Oktober")
    case 11:
        print("November")
    case 12:
        print("Desember")
    case _:
        print("Angka tidak valid")
```

Penjelasan:

1. Baris 1 (bulan_1001 = int(input("masukkan angka bulan (1-12): ")))
Mengambil masukan angka dari pengguna, mengonversinya menjadi bilangan bulat (integer), lalu menyimpannya ke dalam variabel bulan_1001.

2. Baris 3 (match bulan_1001:): Membuka struktur penyeleksian kondisi match-case untuk mengevaluasi nilai yang tersimpan di dalam variabel bulan_1001.

3. Baris 4–27 (case 1: hingga case 12:): Menjodohkan nilai bulan_1001 secara langsung dengan angka 1 sampai 12:

   a. case 1: Mencetak "Januari" 

   b. case 2: Mencetak "Februari"

   c. case 3: Mencetak "Maret"

   d. case 4: Mencetak "April"

   e. case 5: Mencetak "Mei"

   f. case 6: Mencetak "Juni"

   g. case 7: Mencetak "Juli"

   h. case 8: Mencetak "Agustus"

   i. case 9: Mencetak "September"

   j. case 10: Mencetak "Oktober"

   k. case 11: Mencetak "November"
   
   l. case 12: Mencetak "Desember"

4. Baris 28–29 (case _: dan print("Angka tidak valid"))
Berfungsi sebagai kondisi default (wildcard). Karakter garis bawah (_) akan menangkap seluruh nilai masukan yang tidak cocok dengan angka 1 sampai 12 (misalnya angka 0, 13, atau -5) dan mencetak pesan "Angka tidak valid".   

5. Catatan Alur: Match-case bekerja mirip seperti switch-case pada bahasa pemrograman lain. Program hanya mengeksekusi satu cabang case yang nilainya cocok secara persis dengan variabel yang dievaluasi.

```python
print("=== SISTEM LOKET ALPRO ADVENTURE PARK ===")

# 1. Input Data Pengunjung
nama_1001 = input("Masukkan Nama Pengunjung        : ")
umur_1001 = int(input("Input umur anda                 : "))
sim_1001 = input("Apakah Anda Sudah Punya SIM C (y/t): ")[0].lower()

print("\nPilihan Paket Wahana (1-5):")
print("  1. Safari Rimba         (Rp 50,000)")
print("  2. Arung Jeram          (Rp 75,000)")
print("  3. Motor ATV Ekstrim    (Rp 120,000)")
print("  4. Roller Coaster Kilat (Rp 100,000)")
print("  5. All-Access VIP       (Rp 220,000)")

paket_1001 = int(input("Masukkan nomor paket (1-5)      : "))
jumlah_tiket_1001 = int(input("Masukkan jumlah tiket           : "))

# Penerapan IF tunggal untuk validasi kuota tiket
if jumlah_tiket_1001 <= 0:
    print("Peringatan: Kuota tiket tidak valid!")

input_member_1001 = input("Apakah Anda member? (y/t)        : ").strip().lower()
is_member_1001 = input_member_1001 in ["y", "ya"]

input_promo_1001 = input("Apakah kode promo valid? (y/t)  : ").strip().lower()
kode_promo_valid_1001 = input_promo_1001 in ["y", "ya"]

# 2. Pemilihan Wahana Menggunakan match-case
harga_satuan_1001 = 0
nama_wahana_1001 = ""

match paket_1001:
    case 1:
        nama_wahana_1001 = "Wahana Safari Rimba"
        harga_satuan_1001 = 50000
    case 2:
        nama_wahana_1001 = "Wahana Arung Jeram"
        harga_satuan_1001 = 75000
    case 3:
        nama_wahana_1001 = "Wahana Motor ATV Ekstrim"
        harga_satuan_1001 = 120000
    case 4:
        nama_wahana_1001 = "Wahana Roller Coaster Kilat"
        harga_satuan_1001 = 100000
    case 5:
        nama_wahana_1001 = "Wahana All-Access VIP"
        harga_satuan_1001 = 220000
    case _:
        print("\nPaket wahana tidak valid!")
        harga_satuan_1001 = 0

if harga_satuan_1001 > 0 and jumlah_tiket_1001 > 0:
    # 3. Validasi Izin Kendali Wahana Menggunakan if-elif-else
    print("\n--- KELAYAKAN PENGENDARA WAHANA ---")
    if paket_1001 == 3:
        if umur_1001 >= 17 and sim_1001 == 'y':
            print("Status Akses: Anda sudah dewasa dan boleh mengendarai ATV sendiri.")
        elif umur_1001 >= 17 and sim_1001 != 'y':
            print("Status Akses: Anda sudah dewasa tetapi tidak boleh bawa motor ATV (wajib didampingi instruktur).")
        elif umur_1001 < 17 and sim_1001 == 'y':
            print("Status Akses: Identitas tidak valid: Belum cukup umur memiliki SIM.")
        else:
            print("Status Akses: Anda belum cukup umur dan tidak boleh bawa motor ATV.")
    else:
        if umur_1001 >= 10:
            print("Status Akses: Anda memenuhi syarat umur untuk wahana ini.")
        else:
            print("Status Akses: Anda belum cukup umur untuk wahana ini.")

    # 4. Akumulasi Diskon Bertingkat Menggunakan Multi-IF Terpisah
    subtotal_1001 = harga_satuan_1001 * jumlah_tiket_1001
    total_diskon_persen_1001 = 0

    if subtotal_1001 >= 200000:
        total_diskon_persen_1001 += 10  # Diskon Belanja Besar

    if is_member_1001:
        total_diskon_persen_1001 += 5   # Diskon Member

    if kode_promo_valid_1001:
        total_diskon_persen_1001 += 15  # Diskon Voucher Promo

    if jumlah_tiket_1001 >= 5:
        total_diskon_persen_1001 += 5   # Diskon Tambahan Rombongan

    # 5. Menghitung Nominal Diskon dan Total Bayar
    nominal_diskon_1001 = subtotal_1001 * (total_diskon_persen_1001 / 100)
    total_bayar_1001 = subtotal_1001 - nominal_diskon_1001

    # Evaluasi Bonus Audit Menggunakan if-else
    if total_bayar_1001 > 300000:
        catatan_layanan_1001 = "Selamat! Anda berhak mendapatkan Souvenir Gratis."
    else:
        catatan_layanan_1001 = "Terima kasih telah berkunjung."

    # Output Rincian Pembayaran
    print("\n--- Rincian Pembayaran ---")
    print(f"Subtotal Belanja : Rp {subtotal_1001:,.0f}")
    print(f"Total Diskon     : {total_diskon_persen_1001}% (Rp {nominal_diskon_1001:,.0f})")
    print(f"Total Bayar      : Rp {total_bayar_1001:,.0f}")
    print(f"Catatan Layanan  : {catatan_layanan_1001}")

print("Program Selesai")
```

Penjelasan:

1. Input Data Pengunjung & Inisialisasi (Pilar IF Tunggal)

   a. Input Data Utama: Meminta masukan nama, umur, sim (mengambil karakter pertama dan diubah ke huruf kecil), serta jumlah_tiket.

   b. IF Tunggal: Mengecek if jumlah_tiket_1001 <= 0 untuk mencetak peringatan jika jumlah tiket tidak valid, tanpa menggunakan blok else.

   c.Input Diskon & Handling: Mengambil status member dan kode promo menggunakan .strip().lower() yang langsung dikonversi menjadi nilai boolean (True/False) dengan operator in ["y", "ya"].  
   
 2. Pemilihan Wahana (Pilar Match-Case)
 
    a.Menggunakan struktur match paket_1001: untuk mencocokkan pilihan paket (1 sampai 5) secara presisi dengan nama wahana dan harga satuannya.

    b.Wildcard (case _:): Menangani input angka paket di luar 1–5 dengan memberikan pesan error dan menetapkan harga ke 0.

3. Validasi Kelayakan Pengendara (Pilar IF-ELIF-ELSE)

   a.Khusus Paket 3 (Motor ATV): Evaluasi menggunakan gabungan operator logika and dan != pada umur dan status SIM:

     umur >= 17 and sim == 'y': Boleh mengendarai sendiri. 
     
     umur >= 17 and sim != 'y': Harus didampingi instruktur.
     
     umur < 17 and sim == 'y': Status identitas tidak valid.
     
     else: Belum cukup umur.
     
   b. Paket Lainnya: Menggunakan if-else sederhana untuk mengecek batas umur minimal 10 tahun.
   
4. Akumulasi Diskon Bertingkat (Pilar Multi-IF Terpisah)

   a. Menghitung subtotal dari harga_satuan * jumlah_tiket.
   
   b. Menggunakan 4 blok if independen/terpisah (bukan elif) agar setiap kondisi diskon yang terpenuhi bisa diakumulasikan (ditumpuk) ke variabel total_diskon_persen_1001:

     1. Subtotal >= 200.000 = +10%

     2. Status Member = True = +5%

     3. Kode Promo Valid = True = +15%
     
     4. Tiket >= 5  =  +5
     
5. Audit Pembayaran & Output (Pilar IF-ELSE Sederhana)

   a. Menghitung nominal_diskon dan total_bayar.
   
   b. Menggunakan if-else untuk mengevaluasi total_bayar > 300000 guna menentukan pesan layanan (dapat souvenir gratis atau ucapan terima kasih biasa).
   
   c. Mencetak seluruh rincian transaksi dengan format desimal :, .0f untuk memisahkan ribuan (contoh: Rp 240,000).

# KESIMPULAN

1. Struktur percabangan pada Python 
Mulai dari if tunggal, if-else, if-elif-else, multiple if, hingga match-case, masing-masing memiliki fungsi dan konteks penggunaan yang berbeda, sehingga pemilihan bentuk percabangan yang tepat sangat memengaruhi kebenaran logika dan efisiensi suatu program.

2. Pernyataan if tunggal cocok digunakan apabila hanya ada satu kemungkinan aksi yang perlu dijalankan, sedangkan if-else diperlukan apabila terdapat dua kemungkinan hasil yang keduanya perlu ditangani secara eksplisit, seperti pada studi kasus pengecekan kelulusan berdasarkan IPK.

3. Struktur if-elif-else paling tepat digunakan pada kondisi-kondisi yang saling eksklusif dan bertingkat, seperti pada penentuan status kelayakan mengendarai motor dan penentuan grade nilai, karena Python akan berhenti mengevaluasi begitu menemukan kondisi pertama yang bernilai True sehingga lebih efisien.

4. Pendekatan multiple if (beberapa if berdiri sendiri) diperlukan ketika kondisi-kondisi yang dievaluasi tidak saling eksklusif dan memungkinkan lebih dari satu blok kode dieksekusi sekaligus, sebagaimana terlihat pada studi kasus perhitungan diskon belanja di mana beberapa jenis diskon dapat diakumulasikan sekaligus.

5. Struktur match-case memberikan solusi yang lebih ringkas dan mudah dibaca dibandingkan if-elif-else yang panjang, khususnya ketika suatu variabel perlu dicocokkan dengan banyak nilai diskrit, seperti pada studi kasus penentuan nama bulan.

6. Operator logika and berperan penting dalam menggabungkan lebih dari satu syarat menjadi satu kondisi tunggal, sehingga suatu blok kode hanya akan dieksekusi apabila seluruh syarat yang digabungkan terpenuhi secara bersamaan.
