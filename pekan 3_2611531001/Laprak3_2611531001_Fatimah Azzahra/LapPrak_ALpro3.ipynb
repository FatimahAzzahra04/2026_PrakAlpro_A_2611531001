## 2.1 Dasar Teori
Operator adalah simbol yang digunakan untuk mengoperasikan nilai atau variabel (disebut operand) guna menghasilkan suatu nilai baru. Berdasarkan fungsinya, operator dalam Python dapat dikelompokkan sebagai berikut.

a. Operator Aritmatika
Operator aritmatika digunakan untuk melakukan operasi matematis dasar, yaitu penjumlahan (+), pengurangan (-), perkalian (*), pembagian (/), pembagian bulat atau floor division (//), sisa bagi atau modulus (%), dan perpangkatan (**).

b. Operator Assignment (Penugasan)
Operator assignment digunakan untuk memberikan atau memperbarui nilai suatu variabel. Selain tanda sama dengan (=) sebagai assignment biasa, Python juga menyediakan assignment gabungan seperti +=, -=, *=, /=, //=, %=, dan **= yang merupakan bentuk singkat dari operasi aritmatika yang diikuti dengan penugasan ulang ke variabel yang sama.

c. Operator Bitwise
Operator bitwise bekerja pada tingkat representasi biner (bit) dari suatu bilangan bulat. Operator ini meliputi AND (&), OR (|), XOR (^), NOT (~ atau dapat direpresentasikan dengan operasi negatif), geser kiri (<<), dan geser kanan (>>). Operator geser kiri akan mengalikan nilai dengan 2 pangkat jumlah pergeseran, sedangkan geser kanan akan membaginya.

d. Operator Keanggotaan (Membership) dan Identitas
Operator keanggotaan, yaitu in dan not in, digunakan untuk memeriksa apakah suatu nilai terdapat atau tidak terdapat di dalam suatu struktur data seperti list, tuple, atau string. Adapun operator identitas, yaitu is dan is not, digunakan untuk memeriksa apakah dua variabel menunjuk pada objek yang sama persis di memori, bukan sekadar memiliki nilai yang sama seperti pada operator perbandingan (==).

e. Operator Logika (Boolean)
Operator logika digunakan untuk mengombinasikan beberapa ekspresi bernilai boolean (True/False). Operator ini meliputi and (konjungsi, bernilai True jika kedua operand True), or (disjungsi, bernilai True jika salah satu operand True), dan not (negasi, membalik nilai boolean). Operasi XOR (disjungsi eksklusif) yang bernilai True jika kedua nilai berbeda dapat diperoleh melalui operator perbandingan tidak sama dengan (!=) pada dua nilai boolean.

f. Operator Perbandingan (Relasi)
Operator perbandingan digunakan untuk membandingkan dua buah nilai dan menghasilkan nilai boolean. Operator ini meliputi lebih besar dari (>), lebih kecil dari (<), lebih besar sama dengan (>=), lebih kecil sama dengan (<=), sama dengan (==), dan tidak sama dengan (!=). Python juga mendukung perbandingan berantai (chained comparison), misalnya 0 < x < 100, yang secara otomatis dievaluasi sebagai (0 < x) and (x < 100).

- Menghitung Besarnya Diskon
Besarnya diskon dihitung dengan mengalikan: total belanja × persentase diskon. Operator yang digunakan adalah *. Menghitung Rata-rata Harga Barang
- Total belanja dibagi dengan jumlah barang.
Operator / digunakan untuk pembagian. Hasilnya menunjukkan perkiraan harga rata-rata setiap barang.
- Menghitung Sisa Pembagian Barang
Operator % digunakan untuk mendapatkan sisa pembagian jumlah barang dengan 2. Hal ini dapat digunakan untuk mengetahui apakah jumlah barang merupakan bilangan genap atau ganjil. Contoh: 4 % 2 = 0 , 5 % 2 = 1

6. Assignment Pengurangan (-=)

8. Operator Identitas

- is not

- is
list_promo1_1001 is list_promo3_1001 menghasilkan True. Keduanya menunjuk pada objek yang sama.

Intinya:
is → memeriksa apakah objeknya sama
is not → memeriksa apakah objeknya berbeda
== → memeriksa apakah nilainya sama

Contohnya jika semua kondisi terpenuhi:

1111 dalam biner memiliki nilai desimal 15.

status_bit_1001 & 1 digunakan untuk memeriksa bit member. Jika hasilnya 1, berarti status member aktif. Hasil tersebut kemudian diubah menjadi Boolean menggunakan bool().
- Pemeriksaan Promo
status_bit_1001 & 8 digunakan untuk memeriksa apakah bit promo aktif. Jika hasilnya 8, berarti promo tersedia. Jika hasilnya 0, berarti promo tidak tersedia.

12. Bitwise XOR (^)
Operator ^ disebut XOR (Exclusive OR). Digunakan untuk membandingkan dua nilai biner. Bit akan menghasilkan 1 jika kedua bit yang dibandingkan berbeda. Jika kedua bit sama, hasilnya 0.
Pada program:
status_bit_1001 dibandingkan dengan kode referensi 11. Angka 11 dalam biner adalah 1011. Hasil XOR menunjukkan bagian bit yang berbeda antara kode transaksi dan kode referensi.

13. Bitwise Shift Left (<<)
Operator << digunakan untuk menggeser bit ke sebelah kiri. Pada program, status_bit_1001 digeser sebanyak 1 posisi ke kiri. Secara umum, pergeseran satu posisi ke kiri pada bilangan biner setara dengan mengalikan bilangan dengan 2.
Contoh:

3 → 6
