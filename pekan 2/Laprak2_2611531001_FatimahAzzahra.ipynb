## 2.1 Dasar Teori

Dalam bahasa pemrograman Python, variabel adalah suatu nama yang digunakan untuk merujuk pada suatu lokasi memori tempat sebuah nilai disimpan. Python bersifat dynamically typed language, sehingga tipe data suatu variabel tidak perlu dideklarasikan secara eksplisit, melainkan ditentukan secara otomatis berdasarkan nilai yang diberikan pada saat penugasan (assignment). Berikut adalah penjelasan mengenai beberapa tipe data dasar yang digunakan dalam praktikum ini.

1. Tipe Data Numerik

Tipe data numerik pada Python terbagi menjadi tiga jenis utama, yaitu:

a.	Integer (int): bilangan bulat tanpa mengandung angka desimal, contohnya 100.

b.	Float: bilangan pecahan atau bilangan yang mengandung angka di belakang koma, contohnya 10.2345.

c.	Complex: bilangan kompleks yang terdiri dari bagian real dan bagian imajiner yang ditandai dengan huruf j, contohnya 100+3j.

Untuk memeriksa jenis tipe data dari suatu variabel, Python menyediakan fungsi bawaan type() yang akan mengembalikan nama kelas (class) dari tipe data variabel tersebut.

2. Tipe Data String

String merupakan tipe data yang digunakan untuk menyimpan kumpulan karakter atau teks, seperti nama, alamat, atau kalimat. Dalam Python, string dapat dituliskan menggunakan tanda kutip tunggal ('...'), tanda kutip ganda ("..."), maupun tanda kutip tiga ("""...""") yang biasa digunakan untuk menuliskan string dengan format lebih dari satu baris (multiline string).

3. Tipe Data Boolean

Boolean adalah tipe data yang hanya memiliki dua kemungkinan nilai, yaitu True (benar) atau False (salah). Tipe data ini umumnya digunakan sebagai hasil dari suatu ekspresi logika atau perbandingan (comparison), misalnya untuk menentukan apakah suatu nilai lebih besar, lebih kecil, atau sama dengan nilai lain, serta digunakan dalam pengambilan keputusan pada struktur percabangan seperti if.

4. Fungsi input

 Fungsi input() untuk menerima masukan dari pengguna. Perlu diperhatikan bahwa nilai yang dihasilkan oleh fungsi input() selalu bertipe string, sehingga apabila nilai tersebut akan digunakan dalam operasi matematis (seperti perkalian atau penjumlahan), maka nilai tersebut harus dikonversi (type casting) terlebih dahulu ke tipe data yang sesuai menggunakan fungsi seperti int() atau float().

5. Konstanta (Constant)

Konstanta adalah suatu variabel yang nilainya bersifat tetap dan tidak boleh diubah selama program dijalankan. Python secara bawaan sebenarnya tidak memiliki mekanisme khusus untuk membuat konstanta sejati, karena semua variabel pada dasarnya dapat diubah nilainya. Namun, Python menyediakan modul typing dengan tipe Final yang dapat digunakan sebagai penanda (annotation) bahwa suatu variabel dimaksudkan untuk tidak diubah nilainya.

Jika suatu variabel yang ditandai Final diubah, maka type checker (seperti mypy) akan memberikan peringatan, meskipun Python interpreter sendiri tidak akan menghentikan eksekusi program. Konvensi penulisan nama konstanta pada umumnya menggunakan huruf kapital seluruhnya, misalnya PI.





## 2.2 Kode Program

2.2.1 Program "Numerik_2611531001"



```python
#create a variable with integer value.
a_1001=100
print("The type of variable having value",a_1001,"is",type(a_1001))

#acreate a variable with float value.
b_1001=10.2345
print("The type of variable having value", b_1001,"is",type(b_1001))

#acreate a variable with complex value.
c_1001=100+3j
print ("The type of variable having value ",c_1001,"is",type(c_1001))
```

Penjelasan :
Tipe Data Angka:

a_1001 = 100 adalah bilangan bulat (Integer).

b_1001 = 10.2345 adalah bilangan desimal (Float).

c_1001 = 100+3j adalah bilangan kompleks (Complex).

2.2.3 Program "Boolean_2611531001"

```python
is_lulus_1001 = True
is_cumlaude_1001 = True

# Menggunakan Boolean
nilai_1001 = 85
batas_lulus_1001 = 75

#Menentukan nilai Boolean dari kondisi
status_kelulusan_1001 = nilai_1001 >= batas_lulus_1001  # Hasilnya akan True

print("=== Check Kelulusan ===")
print("Nilai:", nilai_1001)
print("Apakah lulus?:", status_kelulusan_1001)
if is_lulus_1001 and is_cumlaude_1001:
    print("Selamat, Anda lulus dengan predikat Cum Laude!")
```

Penjelasan:

1. Inisialisasi Variabel: is_lulus_1001 dan is_cumlaude_1001 bernilai True. nilai_1001 diisi 85 dan batas_lulus_1001 diisi 75.

2. Evaluasi Kondisi: Variabel status_kelulusan_1001 mengecek apakah nilai lebih besar atau sama dengan batas lulus (85 >= 75), yang menghasilkan nilai True.

2.2.4 Program "nama input_2611531001"

```python
panjang_1001= input('Masukkan nilai panjang: ')
lebar_1001 = input('Masukkan nilai lebar: ')
luas_1001 = int(panjang_1001) * int(lebar_1001)
print("Luas =", luas_1001)
```

Penjelasan:
1. Input Data: Menerima masukan dari pengguna untuk nilai panjang dan lebar melalui fungsi input().

2. Konversi & Operasi: Mengonversi input teks menjadi bilangan bulat menggunakan int(), lalu mengalikannya untuk menghitung luas.

2.2.5 Program "Konstanta_2611531001"

```python
from typing import Final
PI: Final = 3.14
print("pI: %f" % (PI))
jari_1001 = float(input('Masukkan nilai jari-jari:'))
luas_1001 = PI * jari_1001 * jari_1001
print("Luas lingkaran dengan jari-jari %.2f adalah %.2f" % (jari_1001, luas_1001))
```

Penjelasan:
1. Mendefinisikan konstanta PI bernilai 3.14 (catatan: kode pada gambar menggunakan terjemahan otomatis "dari pengetikan impor Final" yang aslinya from typing import Final).

2. Input & Operasi: Meminta masukan jari-jari dari pengguna, mengonversinya ke desimal (float), lalu menghitung luas lingkaran (Luas = PI x r**).
